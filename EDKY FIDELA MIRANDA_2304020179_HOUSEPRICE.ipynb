```python
# ==========================================
# 1. Import Library
# ==========================================
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# 2. Upload File CSV
# ==========================================
from google.colab import files
uploaded = files.upload()

# ==========================================
# 3. Baca File
# ==========================================
df = pd.read_csv("housePrice.csv")

# Bersihkan spasi pada nama kolom
df.columns = df.columns.str.strip()

print("Kolom dalam dataset:")
print(df.columns)

# ==========================================
# 4. HISTOGRAM AREA
# ==========================================
plt.figure()
plt.hist(df["Area"], bins=10)
plt.title("Distribusi Area")
plt.xlabel("Area (Square meters)")
plt.ylabel("Frekuensi")
plt.show()

# ==========================================
# 5. HISTOGRAM ROOM
# ==========================================
plt.figure()
plt.hist(df["Room"], bins=6)
plt.title("Distribusi Jumlah Room")
plt.xlabel("Jumlah Room")
plt.ylabel("Frekuensi")
plt.show()

# ==========================================
# 6. PIE CHART PARKING
# ==========================================
plt.figure()
parking_counts = df["Parking"].value_counts()
plt.pie(parking_counts, labels=parking_counts.index, autopct='%1.1f%%')
plt.title("Persentase Parking")
plt.show()

# ==========================================
# 7. PIE CHART WAREHOUSE
# ==========================================
plt.figure()
warehouse_counts = df["Warehouse"].value_counts()
plt.pie(warehouse_counts, labels=warehouse_counts.index, autopct='%1.1f%%')
plt.title("Persentase Warehouse")
plt.show()

# ==========================================
# 8. PIE CHART ELEVATOR
# ==========================================
plt.figure()
elevator_counts = df["Elevator"].value_counts()
plt.pie(elevator_counts, labels=elevator_counts.index, autopct='%1.1f%%')
plt.title("Persentase Elevator")
plt.show()

# ==========================================
# 9. BAR CHART ADDRESS
# ==========================================
plt.figure()
address_counts = df["Address"].value_counts()
plt.bar(address_counts.index, address_counts.values)
plt.title("Distribusi Address")
plt.xlabel("Region")
plt.ylabel("Jumlah Rumah")
plt.xticks(rotation=45)
plt.show()

# ==========================================
# 10. HISTOGRAM PRICE
# ==========================================
plt.figure()
plt.hist(df["Price"], bins=10)
plt.title("Distribusi Price (Toman)")
plt.xlabel("Price")
plt.ylabel("Frekuensi")
plt.show()

# ==========================================
# 11. HISTOGRAM PRICE USD
# ==========================================
plt.figure()
plt.hist(df["Price(USD)"], bins=10)
plt.title("Distribusi Price (USD)")
plt.xlabel("Price (USD)")
plt.ylabel("Frekuensi")
plt.show()
```
