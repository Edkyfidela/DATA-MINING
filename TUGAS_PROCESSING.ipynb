```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# =========================
# Load dataset
# =========================
df = pd.read_csv('Islander_data.csv')

# =========================
# 1. Preview Data
# =========================
print("\n=== 1. PREVIEW DATA ===")

# supaya semua data tampil
pd.set_option('display.max_rows', None)

print(df)
```

Hasil pengecekan menunjukkan jumlah data yang kosong pada setiap kolom. Jika terdapat nilai nol (0), berarti tidak ada data yang hilang, sehingga dataset sudah cukup lengkap untuk dianalisis.

```python
# =========================
# 2. Info Dataset
# =========================
print("\n=== 2. INFO DATASET ===")

# info dasar
print("\n--- Info Dasar ---")
df.info()

# jumlah baris & kolom
print("\n--- Jumlah Data ---")
print(f"Jumlah baris: {df.shape[0]}")
print(f"Jumlah kolom: {df.shape[1]}")

# tipe data
print("\n--- Tipe Data ---")
print(df.dtypes)

# statistik deskriptif
print("\n--- Statistik Deskriptif ---")
print(df.describe())
```

Pada bagian ini berisi:

Info dasar, yaitu menunjukkan jumlah data dan tipe tiap kolom
Jumlah data, yaitu total baris dan kolom
Tipe data, yaitu memastikan apakah data sudah sesuai (angka atau kategori)
Statistik deskriptif, yaitu nilai rata-rata, minimum, maksimum

```python
# =========================
# 3. Missing Value
# =========================
print("\n=== 3. MISSING VALUE ===")
missing = df.isnull().sum()
print(missing)
```

Mengecek apakah ada data yang kosong. Jika ada missing value, perlu di perbaiki agar analisisnya tidak salah.

```python
# =========================
# 4. Cleaning Data
# =========================
print("\n=== 4. DATA SETELAH CLEANING ===")

categorical_cols = ['Happy_Sad_group', 'Drug']

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.lower()

# Semua data
print(df[categorical_cols])
```

Data kategori seperti Happy_Sad_group dan Drug dibersihkan dengan menghapus spasi dan membuat huruf kecil semua. Tujuannya supaya datanya konsisten dan tidak ada perbedaan penulisan.

```python
# =========================
# 5. Feature Engineering
# =========================
df['Diff_Baru'] = df['Mem_Score_After'] - df['Mem_Score_Before']

print("\n=== 5. DATA DIFF BARU ===")

# Semua data
print(df[['Mem_Score_Before','Mem_Score_After','Diff_Baru']])
```

Dengan dibuat kolom baru yaitu Diff_Baru. Yang berisi selisih nilai setelah dan sebelum. Jadi bisa langsung terlihat perubahan kemampuan memorinya.



```python
# =========================
# 6. Smoothing
# =========================
numeric_cols = ['age', 'Mem_Score_Before', 'Mem_Score_After', 'Diff']

for col in numeric_cols:
    if col in df.columns:
        df[col + '_smooth'] = df[col].rolling(window=3).mean()

print("\n=== 6. DATA SETELAH SMOOTHING ===")

smooth_cols = [col for col in df.columns if 'smooth' in col]

# Semua data
print(df[smooth_cols])
```

Smoothing dilakukan dengan rata-rata bergerak. Tujuannya agar data lebih halus dan mengurangi noise.

PERBAIKAN GRAFIK

```python
# =========================
# 7. Histogram Umur
# =========================
plt.figure()

# pastikan numerik
umur = pd.to_numeric(df['age'], errors='coerce').dropna()

plt.hist(umur)
plt.title('Distribusi Umur')
plt.xlabel('Umur')
plt.ylabel('Frekuensi')
plt.show()
```

Grafik menunjukkan distribusi umur responden, di mana terdapat umur yang jumlahnya lebih banyak dibandingkan umur lainnya.

```python
# =========================
# 8. Scatter Plot
# =========================
plt.figure()

# pastikan numerik
df['Mem_Score_Before'] = pd.to_numeric(df['Mem_Score_Before'], errors='coerce')
df['Mem_Score_After'] = pd.to_numeric(df['Mem_Score_After'], errors='coerce')

# hapus data kosong
df_scatter = df.dropna(subset=['Mem_Score_Before', 'Mem_Score_After'])

plt.scatter(df_scatter['Mem_Score_Before'], df_scatter['Mem_Score_After'])
plt.title('Before vs After Memory Score')
plt.xlabel('Memory Score Before')
plt.ylabel('Memory Score After')
plt.show()
```

Grafik menunjukkan perbandingan nilai sebelum dan sesudah, di mana sebagian besar nilai setelah cenderung lebih tinggi.

```python
# =========================
# 9. Boxplot
# =========================
plt.figure()

data_box = df['Mem_Score_Before'].dropna()

sns.boxplot(x=data_box)
plt.title('Boxplot Memory Score Before')
plt.xlabel('Memory Score Before')
plt.show()
```

Grafik menunjukkan persebaran nilai sebelum, terlihat adanya variasi data serta beberapa nilai ekstrem.

```python
# =========================
# 10. Line Plot
# =========================
plt.figure()

data_line = df['Mem_Score_Before'].dropna()

# smoothing
smooth = data_line.rolling(window=3).mean()

plt.plot(data_line, label='Data Asli')
plt.plot(smooth, label='Smoothing')

plt.legend()
plt.title('Perbandingan Data Asli dan Smoothing')
plt.xlabel('Index Data')
plt.ylabel('Memory Score')
plt.show()
```

Grafik menunjukkan perbandingan data asli dengan hasil smoothing,
di mana data smoothing terlihat lebih stabil dibandingkan data aslinya.

```python
# =========================
# 11. Histogram Diff
# =========================
plt.figure()

# gunakan kolom asli Diff kalau ada, kalau tidak buat baru
if 'Diff' in df.columns:
    diff = pd.to_numeric(df['Diff'], errors='coerce')
else:
    diff = df['Mem_Score_After'] - df['Mem_Score_Before']

diff = diff.dropna()

plt.hist(diff)
plt.title('Distribusi Perubahan Nilai')
plt.xlabel('Selisih Nilai')
plt.ylabel('Frekuensi')
plt.show()
```

Grafik menunjukkan perubahan nilai,
di mana sebagian besar data berada di atas nol yang berarti terjadi peningkatan.

```python
# =========================
# 12. Simpan Data
# =========================
df.to_csv('hasil_preprocessing.csv', index=False)

print("\nData berhasil disimpan sebagai 'hasil_preprocessing.csv'")
```
