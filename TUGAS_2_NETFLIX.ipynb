```python
# ==========================================
# GRAFIK 1 - DISTRIBUSI TIPE KONTEN
# ==========================================
plt.figure(figsize=(6,4))
sns.barplot(x=type_counts.index, y=type_counts.values)
plt.title("Distribusi Tipe Konten Netflix")
plt.xlabel("Tipe Konten")
plt.ylabel("Jumlah")
plt.show()

print("""
1️⃣ Grafik Distribusi Tipe Konten (Movie vs TV Show)

Penjelasan:
Grafik ini menunjukkan perbandingan jumlah konten Netflix berdasarkan tipe kontennya, yaitu Movie dan TV Show.
Dari grafik terlihat bahwa jumlah Movie jauh lebih banyak dibandingkan TV Show.
Hal ini menunjukkan bahwa Netflix lebih banyak menyediakan film sebagai bentuk konten utama di platformnya.
Film biasanya lebih mudah diproduksi dan dikonsumsi dalam satu kali tontonan dibandingkan serial televisi.
Oleh karena itu, dominasi Movie pada grafik ini menggambarkan strategi Netflix dalam menyediakan konten yang cepat dikonsumsi oleh penonton.
""")

# ==========================================
# GRAFIK 2 - PIE CHART MOVIE VS TV SHOW
# ==========================================
plt.figure(figsize=(6,6))
plt.pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%')
plt.title("Persentase Movie vs TV Show")
plt.show()

print("""
2️⃣ Grafik Persentase Movie vs TV Show (Pie Chart)

Penjelasan:
Grafik pie ini menunjukkan persentase distribusi konten Netflix antara Movie dan TV Show.
Berdasarkan grafik tersebut, terlihat bahwa sebagian besar konten yang tersedia di Netflix adalah Movie,
sedangkan TV Show memiliki persentase yang lebih kecil.
Hal ini memperkuat hasil dari grafik sebelumnya bahwa film merupakan konten yang paling dominan.
Persentase ini menunjukkan bahwa Netflix lebih banyak berinvestasi dalam produksi atau distribusi film dibandingkan serial televisi.
Dengan demikian, dapat disimpulkan bahwa film merupakan jenis konten utama dalam katalog Netflix.
""")

# ==========================================
# GRAFIK 3 - TREND KONTEN PER TAHUN
# ==========================================
plt.figure(figsize=(8,5))
plt.plot(year_counts.index, year_counts.values)
plt.title("Jumlah Konten Berdasarkan Tahun")
plt.xlabel("Tahun")
plt.ylabel("Jumlah Konten")
plt.show()

print("""
3️⃣ Grafik Tren Jumlah Konten Berdasarkan Tahun Rilis

Penjelasan:
Grafik ini menggambarkan perkembangan jumlah konten Netflix berdasarkan tahun rilisnya.
Terlihat bahwa jumlah konten mengalami peningkatan yang signifikan terutama setelah tahun 2000.
Peningkatan ini menunjukkan bahwa industri hiburan digital berkembang pesat dalam beberapa dekade terakhir.
Selain itu, Netflix juga semakin aktif menambahkan berbagai konten baru untuk menarik lebih banyak pengguna.
Tren ini menunjukkan bahwa produksi dan distribusi konten hiburan semakin meningkat seiring dengan perkembangan teknologi streaming.
""")

# ==========================================
# GRAFIK 4 - TOP 10 NEGARA
# ==========================================
plt.figure(figsize=(8,5))
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title("Top 10 Negara dengan Konten Netflix Terbanyak")
plt.xlabel("Jumlah Konten")
plt.ylabel("Negara")
plt.show()

print("""
4️⃣ Grafik Top 10 Negara dengan Konten Netflix Terbanyak

Penjelasan:
Grafik ini menampilkan sepuluh negara yang memiliki jumlah konten Netflix terbanyak.
Dari grafik tersebut terlihat bahwa Amerika Serikat menempati posisi pertama dengan jumlah konten yang paling tinggi.
Hal ini wajar karena Amerika Serikat merupakan salah satu pusat industri perfilman terbesar di dunia, terutama melalui Hollywood.
Negara lain seperti India, Inggris, dan Jepang juga memiliki kontribusi yang cukup besar terhadap konten Netflix.
Grafik ini menunjukkan bahwa produksi konten Netflix berasal dari berbagai negara, meskipun masih didominasi oleh Amerika Serikat.
""")

# ==========================================
# GRAFIK 5 - DISTRIBUSI RATING
# ==========================================
plt.figure(figsize=(8,5))
sns.barplot(x=rating_counts.index, y=rating_counts.values)
plt.xticks(rotation=45)
plt.title("Distribusi Rating Konten")
plt.xlabel("Rating")
plt.ylabel("Jumlah Konten")
plt.show()

print("""
5️⃣ Grafik Distribusi Rating Konten

Penjelasan:
Grafik ini menunjukkan distribusi rating usia dari konten yang tersedia di Netflix.
Rating seperti TV-MA dan TV-14 biasanya memiliki jumlah yang paling banyak dibandingkan rating lainnya.
Hal ini menunjukkan bahwa sebagian besar konten Netflix ditujukan untuk remaja dan penonton dewasa.
Konten dengan rating anak-anak seperti TV-Y atau TV-G jumlahnya relatif lebih sedikit.
Dari grafik ini dapat disimpulkan bahwa target utama penonton Netflix adalah kelompok usia remaja hingga dewasa.
""")

# ==========================================
# GRAFIK 6 - DISTRIBUSI DURASI FILM
# ==========================================
plt.figure(figsize=(8,5))
sns.histplot(df['duration_int'].dropna(), bins=30)
plt.title("Distribusi Durasi Film")
plt.xlabel("Durasi (menit)")
plt.ylabel("Jumlah Film")
plt.show()

print("""
6️⃣ Grafik Distribusi Durasi Film

Penjelasan:
Grafik histogram ini menunjukkan sebaran durasi film yang tersedia di Netflix.
Dari grafik terlihat bahwa sebagian besar film memiliki durasi sekitar 80 hingga 120 menit.
Durasi tersebut merupakan durasi standar untuk film pada umumnya.
Film dengan durasi sangat pendek atau sangat panjang jumlahnya relatif lebih sedikit.
Hal ini menunjukkan bahwa mayoritas film di Netflix mengikuti standar durasi film yang umum dalam industri perfilman.
""")

# ==========================================
# GRAFIK 7 - HEATMAP KORELASI
# ==========================================
plt.figure(figsize=(6,4))
corr = df[['release_year','duration_int']].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Heatmap Korelasi")
plt.show()

print("""
7️⃣ Grafik Heatmap Korelasi

Penjelasan:
Heatmap ini menunjukkan hubungan korelasi antara variabel numerik dalam dataset,
yaitu tahun rilis dan durasi film.
Nilai korelasi yang mendekati 0 menunjukkan bahwa hubungan antara kedua variabel tersebut sangat lemah.
Artinya, panjang atau pendeknya durasi film tidak dipengaruhi oleh tahun rilisnya.
Film yang dirilis pada tahun terbaru tidak selalu memiliki durasi lebih panjang dibandingkan film lama.
Dengan demikian, tidak terdapat hubungan yang signifikan antara tahun rilis film dengan durasi film pada dataset ini.
""")
```
